```python
template_prompt = """ isso é um template de prompt """
```

```python
def chain() -> RunnableSequence:

    prompt = ChatPromptTemplate(
        message = [
            ("system", template_prompt),
            ("placeholder", "{chat_history}"),
            ("human", "{input}")
        ]
        input_variables = ["input", "chat_history"]
    )

    tochain = prompt | CHAT 

    return tochain
```
